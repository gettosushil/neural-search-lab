Recreate and load the trained model

````python
import torch.nn as nn
import torch.nn.functional as F

class TextEncoder(nn.Module):

    def __init__(self, vocab_size, embedding_dim=32, output_dim=16):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            embedding_dim
        )

        self.projection = nn.Linear(
            embedding_dim,
            output_dim
        )

    def forward(self, token_ids):

        # [batch, tokens, 32]
        word_vectors = self.embedding(token_ids)

        # [batch, 32]
        sentence_vector = word_vectors.mean(dim=1)

        # [batch, 16]
        output = self.projection(sentence_vector)

        # Normalize to length 1
        return F.normalize(output, dim=1)


# Reconstruct model
model = TextEncoder(
    vocab_size=len(vocab),
    embedding_dim=checkpoint["embedding_dim"],
    output_dim=checkpoint["output_dim"]
)

# Load our trained weights
model.load_state_dict(checkpoint["model_state"])

model.train()

print(model)
````

# For this controlled experiment, I actually want to fail loudly if a word doesn't exist instead of silently mapping it to <UNK>.

````python
def to_ids(text):

    tokens = tokenize(text)

    ids = []

    for token in tokens:

        if token not in vocab:
            raise ValueError(
                f"Token '{token}' is not in vocabulary"
            )

        ids.append(vocab[token])

    return tokens, ids
````

## Prove the SGD equation .Because we're using: lr = 0.01. SGD performs:

## Wnew ​= Wold ​− 0.01×gradient

## Let's verify dimension 0 manually.

Expected and Actual should match apart from tiny floating-point differences.

That gives you direct experimental proof of:

weightnew = weightold − learningRate × gradient

````python
token = "neural"
token_id = vocab[token]

old_value = embedding_before[token_id][0]

gradient = model.embedding.weight.grad[token_id][0]

expected_new_value = (
    old_value - 0.01 * gradient
)

actual_new_value = (
    model.embedding.weight[token_id][0].detach()
)

print("Old value:")
print(old_value.item())

print("\nGradient:")
print(gradient.item())

print("\nExpected:")
print(expected_new_value.item())

print("\nActual:")
print(actual_new_value.item())
````

### Let's create a serach utility

````python
# ============================================================
# SEMANTIC SEARCH UTILITY — TOP 3 RESULTS
# ============================================================

import sys
import json
from pathlib import Path

import torch
import torch.nn.functional as F


# ============================================================
# 1. PROJECT SETUP
# ============================================================

PROJECT_ROOT = Path("..").resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.model import TextEncoder


MODEL_PATH = PROJECT_ROOT / "models" / "text_encoder.pt"
DATA_PATH = PROJECT_ROOT / "data" / "training_pairs.json"


# ============================================================
# 2. LOAD TRAINED MODEL CHECKPOINT
# ============================================================

checkpoint = torch.load(
    MODEL_PATH,
    map_location="cpu",
    weights_only=True
)

vocab = checkpoint["vocab"]


# ============================================================
# 3. RECREATE MODEL ARCHITECTURE
# ============================================================

model = TextEncoder(
    vocab_size=len(vocab),
    embedding_dim=checkpoint["embedding_dim"],
    output_dim=checkpoint["output_dim"]
)


# Load trained weights
model.load_state_dict(
    checkpoint["model_state"]
)


# Put model into inference mode
model.eval()


print("Model loaded successfully")
print("Vocabulary size :", len(vocab))
print("Embedding dim   :", checkpoint["embedding_dim"])
print("Output dim      :", checkpoint["output_dim"])


# ============================================================
# 4. LOAD DOCUMENTS
# ============================================================

with open(
    DATA_PATH,
    "r",
    encoding="utf-8"
) as f:

    data = json.load(f)


documents = data["documents"]


print("\nDocuments available for search:", len(documents))


# ============================================================
# 5. TEXT -> TOKEN TENSOR
# ============================================================

def text_to_tensor(text):

    tokens = tokenize(text)

    # Try to find UNK token
    unk_id = None

    if "<UNK>" in vocab:
        unk_id = vocab["<UNK>"]

    elif "<unk>" in vocab:
        unk_id = vocab["<unk>"]


    token_ids = []


    for token in tokens:

        if token in vocab:

            token_ids.append(
                vocab[token]
            )

        elif unk_id is not None:

            token_ids.append(
                unk_id
            )

        else:

            # Ignore words that are completely unknown
            print(
                f"Warning: ignoring unknown token: '{token}'"
            )


    # Nothing usable in the query
    if len(token_ids) == 0:
        return None


    return torch.tensor(
        [token_ids],
        dtype=torch.long
    )


# ============================================================
# 6. TEXT -> 16-D VECTOR
# ============================================================

def encode_text(text):

    tensor = text_to_tensor(text)

    if tensor is None:
        return None


    with torch.no_grad():

        vector = model(tensor)


    return vector


# ============================================================
# 7. PRECOMPUTE DOCUMENT VECTORS
# ============================================================

document_vectors = []


for document in documents:

    document_id = document["id"]

    document_text = document["text"]


    vector = encode_text(
        document_text
    )


    if vector is not None:

        document_vectors.append({

            "id": document_id,

            "text": document_text,

            "vector": vector
        })


print(
    "Documents successfully encoded:",
    len(document_vectors)
)


# ============================================================
# 8. SEARCH FUNCTION
# ============================================================

def search(question, top_k=3):

    # --------------------------------------------------------
    # Encode user question
    # --------------------------------------------------------

    query_vector = encode_text(
        question
    )


    if query_vector is None:

        print(
            "\nCould not encode query."
        )

        return []


    # --------------------------------------------------------
    # Display query internals
    # --------------------------------------------------------

    query_tokens = tokenize(
        question
    )


    print("\nQUERY TOKENS:")
    print(query_tokens)


    print("\nTOKEN IDs:")


    for token in query_tokens:

        if token in vocab:

            print(
                f"{token:20s} -> {vocab[token]}"
            )

        else:

            print(
                f"{token:20s} -> OOV"
            )


    print("\nFINAL 16-D QUERY VECTOR:")

    print(
        query_vector.squeeze(0)
    )


    # --------------------------------------------------------
    # Calculate similarity against every document
    # --------------------------------------------------------

    results = []


    for document in document_vectors:

        similarity = F.cosine_similarity(

            query_vector,

            document["vector"]

        ).item()


        results.append({

            "id": document["id"],

            "text": document["text"],

            "similarity": similarity
        })


    # --------------------------------------------------------
    # Sort highest similarity first
    # --------------------------------------------------------

    results.sort(

        key=lambda x: x["similarity"],

        reverse=True
    )


    return results[:top_k]


# ============================================================
# 9. INTERACTIVE SEARCH
# ============================================================

question = input(
    "\nEnter your question: "
)


results = search(
    question,
    top_k=3
)


# ============================================================
# 10. DISPLAY TOP 3 RESULTS
# ============================================================

print("\n")
print("=" * 80)

print(
    "TOP 3 SEMANTIC SEARCH RESULTS"
)

print("=" * 80)


if not results:

    print(
        "No results found."
    )

else:

    for rank, result in enumerate(
        results,
        start=1
    ):

        print(
            f"\nRank #{rank}"
        )

        print(
            f"Document ID : "
            f"{result['id']}"
        )

        print(
            f"Similarity  : "
            f"{result['similarity']:.6f}"
        )

        print(
            f"Answer      : "
            f"{result['text']}"
        )

        print(
            "-" * 80
        )
````
