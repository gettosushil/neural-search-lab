```python
import torch

checkpoint = torch.load(
    "../models/text_encoder.pt",
    map_location="cpu",
    weights_only=True,
)

vocab = checkpoint["vocab"]

initial = checkpoint["initial_embeddings"]

trained = checkpoint["model_state"]["embedding.weight"]

word = "neural"
word_id = vocab[word]

print("Word:", word)
print("Token ID:", word_id)

print("\nInitial word vector:")
print(initial[word_id])

print("\nTrained word vector:")
print(trained[word_id])

print("\nChange in weights:")
print(trained[word_id] - initial[word_id])
```
